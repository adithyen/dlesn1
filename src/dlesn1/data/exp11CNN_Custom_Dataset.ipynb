```python
import tensorflow as tf
```

```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense
from tensorflow.keras.preprocessing.image import ImageDataGenerator
import matplotlib.pyplot as plt
import os
```

```python
base_dir = r"C:\Users\cs2\Desktop\"
img_size = 124

datagen = ImageDataGenerator(
    rescale=1./255,
    validation_split=0.2,
    shear_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True
)
```

```python
train_gen = datagen.flow_from_directory(
    base_dir,
    target_size=(img_size, img_size),
    batch_size=32,
    class_mode='binary',
    subset='training',
    shuffle=True
)
```

```python
val_gen = datagen.flow_from_directory(
    base_dir,
    target_size=(img_size, img_size),
    batch_size=32,
    class_mode='binary',
    subset='validation',
    shuffle=True
)
```

```python
model = Sequential([
    Conv2D(6, kernel_size=(5, 5), activation='relu', input_shape=(img_size, img_size, 3)),
    MaxPooling2D(pool_size=(2, 2)),
    Conv2D(10, kernel_size=(5, 5), activation='relu'),
    MaxPooling2D(pool_size=(2, 2)),
    Flatten(),
    Dense(120, activation='relu'),
    Dense(84, activation='relu'),
    Dense(1, activation='sigmoid')
])
```

```python
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
history = model.fit(train_gen, validation_data=val_gen, epochs=5)
```

```python
loss, acc = model.evaluate(val_gen)
print(f"Validation Accuracy: {acc*100:.2f}%")
```

```python
x_batch, y_batch = next(val_gen)
preds = model.predict(x_batch)
```

```python
x_batch, y_batch = next(val_gen)
preds = model.predict(x_batch)

plt.figure(figsize=(15, 6))
for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(x_batch[i].reshape(img_size, img_size, 3))
    pred_label = 'Dog' if preds[i] > 0.5 else 'Cat'
    actual_label = 'Dog' if y_batch[i] == 1 else 'Cat'
    plt.title(f"Pred: {pred_label}\nActual: {actual_label}")
    plt.axis('off')
plt.tight_layout()
plt.show()
```
